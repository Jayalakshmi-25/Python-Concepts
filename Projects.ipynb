# Home Credit Default Risk – application_train Dashboard
## 20 Streamlit Page Views

This dashboard is designed for the **Home Credit Default Risk – application_train.csv** dataset.

The main objective is to analyze loan applicants, understand customer characteristics, study credit-risk patterns, and identify factors associated with loan repayment/default.

---

# Dataset Overview

Important columns commonly available in `application_train.csv`:

## Identification & Target

- SK_ID_CURR
- TARGET

### TARGET Meaning

- TARGET = 0 → Customer repaid / No payment difficulty
- TARGET = 1 → Customer had payment difficulties / Higher default risk


## Customer Information

- CODE_GENDER
- DAYS_BIRTH
- CNT_CHILDREN
- CNT_FAM_MEMBERS
- NAME_FAMILY_STATUS
- NAME_EDUCATION_TYPE
- NAME_INCOME_TYPE
- NAME_HOUSING_TYPE
- OCCUPATION_TYPE
- ORGANIZATION_TYPE


## Loan Information

- AMT_INCOME_TOTAL
- AMT_CREDIT
- AMT_ANNUITY
- AMT_GOODS_PRICE
- NAME_CONTRACT_TYPE


## Employment Information

- DAYS_EMPLOYED
- DAYS_REGISTRATION
- DAYS_ID_PUBLISH
- OWN_CAR_AGE


## External Credit Scores

- EXT_SOURCE_1
- EXT_SOURCE_2
- EXT_SOURCE_3


## Asset Information

- FLAG_OWN_CAR
- FLAG_OWN_REALTY


## Region Information

- REGION_POPULATION_RELATIVE
- REGION_RATING_CLIENT
- REGION_RATING_CLIENT_W_CITY
- REG_REGION_NOT_LIVE_REGION
- REG_REGION_NOT_WORK_REGION
- REG_CITY_NOT_LIVE_CITY
- REG_CITY_NOT_WORK_CITY

# Main Dashboard KPI Section

| KPI | Calculation |
|---|---|
| Total Applications | Number of Rows |
| Default Customers | TARGET = 1 |
| Non-Default Customers | TARGET = 0 |
| Default Rate | Defaults / Applications × 100 |
| Average Income | Mean AMT_INCOME_TOTAL |
| Average Credit | Mean AMT_CREDIT |
| Average Annuity | Mean AMT_ANNUITY |
| Average Age | abs(DAYS_BIRTH) / 365 |


# Important Derived Features

## 1. Age

Age = abs(DAYS_BIRTH) / 365


## 2. Employment Years

Employment Years = abs(DAYS_EMPLOYED) / 365


## 3. Credit-to-Income Ratio

Credit Income Ratio =

AMT_CREDIT / AMT_INCOME_TOTAL


## 4. Annuity-to-Income Ratio

Annuity Income Ratio =

AMT_ANNUITY / AMT_INCOME_TOTAL


## 5. Credit-to-Goods Ratio

Credit Goods Ratio =

AMT_CREDIT / AMT_GOODS_PRICE


## 6. Average External Score

Average External Score =

mean(
    EXT_SOURCE_1,
    EXT_SOURCE_2,
    EXT_SOURCE_3
)
